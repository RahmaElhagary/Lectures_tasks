```python
from sklearn.datasets import load_wine
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
```

```python
wine = load_wine()
X_train, X_test, y_train, y_test = train_test_split(
    wine.data, wine.target, test_size=0.2, random_state=42
)
```

```python
with mlflow.start_run(run_name="log_reg"):
    lr = LogisticRegression(max_iter=2000)
    lr.fit(X_train, y_train)
    preds = lr.predict(X_test)
    acc = accuracy_score(y_test, preds)

    mlflow.log_param("model", "LogisticRegression")
    mlflow.log_metric("accuracy", acc)
    print(f"LOG_REG Accuracy: {acc}")
```

```python
estimators_list = [1, 20, 60]
for n in estimators_list:
    with mlflow.start_run(run_name="Random_forset"):
        rf = RandomForestClassifier(n_estimators=n,random_state=42)
        rf.fit(X_train, y_train)
        preds = rf.predict(X_test)
        acc = accuracy_score(y_test, preds)

        mlflow.log_param("model", "RandomForest")
        mlflow.log_param("n_estimators", n)
        mlflow.log_metric("accuracy", acc)
        print(f"RFC (n={n}) Accuracy: {acc}")
        
```

```python
kernels_list = ["poly", "rbf", "linear"]
for k in kernels_list:
  with mlflow.start_run(run_name=f"SVM_{k}"):
    svm_model = SVC(kernel=k, probability=True)
    svm_model.fit(X_train, y_train)
    preds = svm_model.predict(X_test)
    acc = accuracy_score(y_test, preds)

    mlflow.log_param("model", "SVM")
    mlflow.log_param("kernel", k)
    mlflow.log_metric("accuracy", acc)

    if k == "rbf":
      mlflow.sklearn.log_model(svm_model, "svm_rbf_model")
      print("-> Saved SVM with RBF model to MLflow!")

    print(f"SVM (kernel={k}) Accuracy: {acc}")
```
