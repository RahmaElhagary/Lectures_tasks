## Task 1

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

(X_train, y_train), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

class_names = [
    'T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]
```

```python
print("=== 1. Data Shapes ===")
print(f"Training images shape: {X_train.shape}")
print(f"Training labels shape: {y_train.shape}")
print(f"Testing images shape:  {X_test.shape}")
print(f"Testing labels shape:  {y_test.shape}\n")

print("=== 2. Classes Info ===")
print(f"Number of classes: {len(class_names)}")
print(f"Class names: {class_names}\n")


plt.figure(figsize=(12, 5))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i], cmap='gray')
    plt.title(f"{y_train[i]}: {class_names[y_train[i]]}")
    plt.axis('off')
plt.suptitle("First 10 Training Images", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
print("=== 4. Class Distribution in Training Set ===")
unique_labels, counts = np.unique(y_train, return_counts=True)
for label, count in zip(unique_labels, counts):
    print(f"Class {label} ({class_names[label]}): {count} images")
```

## Answers 
- Yes the data is balanced
- We do normalization to make all features in a uniform scale for better trainnig

## Task 2

```python
import mlflow
import mlflow.tensorflow
import tensorflow as tf
from keras import layers, models
```

```python
print(mlflow.get_tracking_uri())
```

```python
mlflow.set_experiment("Fashion_MNIST_Baseline")

X_train = X_train / 255.0
X_test = X_test / 255.0

with mlflow.start_run(run_name="Baseline_Model_Run"):
    mlflow.tensorflow.autolog()

    model = models.Sequential([
        layers.Flatten(input_shape=(28, 28)),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    history = model.fit(
        X_train, y_train,
        epochs=10,
        batch_size=64,
        validation_split=0.2
    )
    
    
    test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
    mlflow.log_metric("test_accuracy", test_acc)
    mlflow.log_metric("test_loss", test_loss)

    print(f"\n=== Final Accuracy Reports ===")
    print(f"Test Accuracy: {test_acc * 100:.2f}%")

    fig, ax = plt.subplots(1, 2, figsize=(12, 4))

    ax[0].plot(history.history['loss'], label='Training Loss')
    ax[0].plot(history.history['val_loss'], label='Validation Loss')
    ax[0].set_title('Loss vs. Validation Loss')
    ax[0].set_xlabel('Epoch')
    ax[0].set_ylabel('Loss')
    ax[0].legend()
    ax[0].grid(True)

    ax[1].plot(history.history['accuracy'], label='Training Accuracy')
    ax[1].plot(history.history['val_accuracy'], label='Validation Accuracy')
    ax[1].set_title('Accuracy vs. Validation Accuracy')
    ax[1].set_xlabel('Epoch')
    ax[1].set_ylabel('Accuracy')
    ax[1].legend()
    ax[1].grid(True)

    plt.tight_layout()
    
    plot_path = "accuracy_loss_plot.png"
    plt.savefig(plot_path)
    mlflow.log_artifact(plot_path)
    plt.show()
```

## Answers 
- Flatten: Transforms multi-dimensional input images into a 1D vector 
- ReLU : Introduces non-linearity into the network max(0,x)
- Softmax: Converts raw logit outputs from the final layer into a probability for all targets

## Task 3

```python
predictions = model.predict(X_test)
predicted_classes = np.argmax(predictions, axis=1)


correct_indices = np.where(predicted_classes == y_test)[0]
incorrect_indices = np.where(predicted_classes != y_test)[0]

selected_indices = [
    correct_indices[0],   
    correct_indices[1],    
    incorrect_indices[0]   
]

plt.figure(figsize=(12, 4))

for i, idx in enumerate(selected_indices):
    image = X_test[idx]
    true_label = class_names[y_test[idx]]
    pred_label_idx = predicted_classes[idx]
    pred_label = class_names[pred_label_idx]
    confidence = predictions[idx][pred_label_idx] * 100
    
    plt.subplot(1, 3, i + 1)
    plt.imshow(image, cmap='gray')
    
    
    title_color = 'green' if pred_label_idx == y_test[idx] else 'red'
    
    plt.title(
        f"True: {true_label}\nPred: {pred_label} ({confidence:.1f}%)",
        color=title_color,
        fontsize=11
    )
    plt.axis('off')

plt.tight_layout()
plt.show()
```

## Answers
- items like shirts and t-shirts share nearly identical shapes.
- the forward pass, the image is flattened into a vector, processed by a hidden layer with ReLU activation, and mapped to 10 class probabilities via Softmax.


## Task 4

```python
def build_model():
    model = models.Sequential([
        layers.Flatten(input_shape=(28, 28)),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model


epoch_list = [5, 10, 20]
results = {}

mlflow.set_experiment("Fashion_MNIST_Epoch_Comparison")

for ep in epoch_list:
    print(f"\n================ Model with {ep} Epochs ================")
    
    with mlflow.start_run(run_name=f"Run_{ep}_Epochs"):
        mlflow.tensorflow.autolog()
        
        model = build_model()
        
        history = model.fit(
            X_train, y_train,
            epochs=ep,
            batch_size=64,
            validation_split=0.2
        )
        
        results[ep] = history.history
        
        test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
        best_val_acc = max(history.history['val_accuracy'])
        
        mlflow.log_metric("best_val_accuracy", best_val_acc)
        mlflow.log_metric("test_accuracy", test_acc)
        
        print(f"Epochs: {ep} | Best Val Accuracy: {best_val_acc * 100:.2f}% | Test Accuracy: {test_acc * 100:.2f}%")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ep, hist in results.items():
    epochs_range = range(1, ep + 1)
    
    axes[0].plot(epochs_range, hist['val_loss'], label=f'{ep} Epochs (Val Loss)')

    axes[1].plot(epochs_range, hist['val_accuracy'], label=f'{ep} Epochs (Val Acc)')

axes[0].set_title('Validation Loss Across Runs')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend()
axes[0].grid(True)

axes[1].set_title('Validation Accuracy Across Runs')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

## Answers
- 20-epoch run

## Task 5

```python
from keras.callbacks import EarlyStopping

def build_model():
    model = models.Sequential([
        layers.Flatten(input_shape=(28, 28)),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

runs_config = {
    "Run_A_No_EarlyStopping": None,
    "Run_B_Patience_3": EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True),
    "Run_C_Patience_5": EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
}

mlflow.set_experiment("Fashion_MNIST_EarlyStopping_Experiment")

for run_name, es_callback in runs_config.items():
    print(f"\n================ Running: {run_name} ================")
    
    with mlflow.start_run(run_name=run_name):
        mlflow.tensorflow.autolog()
        
        model = build_model()
        callbacks_list = [es_callback] if es_callback else []
        
        history = model.fit(
            X_train, y_train,
            epochs=40,
            batch_size=64,
            validation_split=0.2,
            callbacks=callbacks_list
        )
        
        actual_epochs = len(history.history['loss'])
        best_val_loss = min(history.history['val_loss'])
        final_val_acc = history.history['val_accuracy'][-1]
        
        mlflow.log_metric("actual_epochs_trained", actual_epochs)
        mlflow.log_metric("best_val_loss", best_val_loss)
        mlflow.log_metric("final_val_accuracy", final_val_acc)
        
        print(f"Actual Epochs Trained: {actual_epochs}")
        print(f"Best Validation Loss:  {best_val_loss:.4f}")
        print(f"Final Validation Acc:  {final_val_acc * 100:.2f}%")
```

## Answers 
- how many consecutive epochs the model is allowed to train without improvement
- EarlyStopping prevents overfitting

## Task 6

```python
import time

neurons_list = [16, 32, 128]

mlflow.set_experiment("Fashion_MNIST_Hidden_Neurons_Experiment")

for n_units in neurons_list:
    print(f"\n================ Training Model with {n_units} Neurons ================")
    
    with mlflow.start_run(run_name=f"Run_{n_units}_Neurons"):
        mlflow.tensorflow.autolog()
        
        model = models.Sequential([
            layers.Flatten(input_shape=(28, 28)),
            layers.Dense(n_units, activation='relu'),
            layers.Dense(10, activation='softmax')
        ])
        
        model.compile(
            optimizer='adam',
            loss='sparse_categorical_crossentropy',
            metrics=['accuracy']
        )
        
        start_time = time.time()
        
        history = model.fit(
            X_train, y_train,
            epochs=10,
            batch_size=64,
            validation_split=0.2
        )
        
        training_time = time.time() - start_time
        
        final_train_acc = history.history['accuracy'][-1]
        final_val_acc = history.history['val_accuracy'][-1]
        
        mlflow.log_metric("training_time_seconds", training_time)
        mlflow.log_metric("final_train_accuracy", final_train_acc)
        mlflow.log_metric("final_val_accuracy", final_val_acc)
        
        print(f"Neurons: {n_units:3d} | Train Acc: {final_train_acc * 100:.2f}% | Val Acc: {final_val_acc * 100:.2f}% | Time: {training_time:.2f}s")
```

## Answers 
- improves validation accuracy up to a point, then overfitting
- The trade-off is that more neurons increase model complexity and training time, raising the risk of overfitting

## Task 7

```python
def build_model():
    model = models.Sequential([
        layers.Flatten(input_shape=(28, 28)),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

batch_sizes = [8, 32, 128]
histories = {}

mlflow.set_experiment("Fashion_MNIST_Batch_Size_Experiment")

for bs in batch_sizes:
    print(f"\n================ Training Model with Batch Size {bs} ================")
    
    with mlflow.start_run(run_name=f"Run_Batch_Size_{bs}"):
        mlflow.tensorflow.autolog()
        
        model = build_model()
        
        start_time = time.time()
        
        history = model.fit(
            X_train, y_train,
            epochs=10,
            batch_size=bs,
            validation_split=0.2
        )
        
        total_time = time.time() - start_time
        time_per_epoch = total_time / 10
        histories[bs] = history.history
        
        final_val_acc = history.history['val_accuracy'][-1]
        
        mlflow.log_metric("time_per_epoch_seconds", time_per_epoch)
        mlflow.log_metric("final_val_accuracy", final_val_acc)
        
        print(f"Batch Size: {bs:3d} | Val Acc: {final_val_acc * 100:.2f}% | Time/Epoch: {time_per_epoch:.2f}s")

plt.figure(figsize=(12, 5))

for bs, hist in histories.items():
    plt.plot(hist['loss'], label=f'Train Loss (BS={bs})')
    plt.plot(hist['val_loss'], linestyle='--', label=f'Val Loss (BS={bs})')

plt.title('Loss Curves for Different Batch Sizes')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Answers 
- Smaller batch sizes create more weight updates per epoch which slows down epoch speed
- Batch size 32

## Task 8

```python
from keras import optimizers

def build_model():
    return models.Sequential([
        layers.Flatten(input_shape=(28, 28)),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])

opt_configs = {
    "SGD_basic": optimizers.SGD(learning_rate=0.01),
    "SGD_momentum": optimizers.SGD(learning_rate=0.01, momentum=0.9),
    "Adam": optimizers.Adam(learning_rate=0.001)
}

histories = {}

mlflow.set_experiment("Fashion_MNIST_Optimizer_Comparison")

for name, opt in opt_configs.items():
    print(f"\n================ Training Model with {name} ================")
    
    with mlflow.start_run(run_name=f"Run_{name}"):
        mlflow.tensorflow.autolog()
        
        model = build_model()
        model.compile(
            optimizer=opt,
            loss='sparse_categorical_crossentropy',
            metrics=['accuracy']
        )
        
        history = model.fit(
            X_train, y_train,
            epochs=10,
            batch_size=64,
            validation_split=0.2
        )
        
        histories[name] = history.history
        val_acc = history.history['val_accuracy'][-1]
        
        print(f"Optimizer: {name} | Final Val Accuracy: {val_acc * 100:.2f}%")

plt.figure(figsize=(12, 5))

for name, hist in histories.items():
    plt.plot(hist['loss'], label=f'{name} (Train Loss)')
    plt.plot(hist['val_loss'], linestyle='--', label=f'{name} (Val Loss)')

plt.title('Training & Validation Loss Curves by Optimizer')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Answers 
- Adam improved the fastest and achieved the highest validation accuracy within 10 epochs due to its adaptive learning rates. Momentum differs from basic SGD by adding a fraction of the previous gradient update to accelerate training along consistent directions and break out of local minima.

## Task 9

```python
activations = ['relu', 'tanh', 'gelu']
histories = {}

mlflow.set_experiment("Fashion_MNIST_Activation_Comparison")

for act in activations:
    print(f"\n================ Training Model with '{act}' Activation ================")
    
    with mlflow.start_run(run_name=f"Run_Activation_{act}"):
        mlflow.tensorflow.autolog()
        
        model = models.Sequential([
            layers.Flatten(input_shape=(28, 28)),
            layers.Dense(128, activation=act),
            layers.Dense(10, activation='softmax')
        ])
        
        model.compile(
            optimizer='adam',
            loss='sparse_categorical_crossentropy',
            metrics=['accuracy']
        )
        
        history = model.fit(
            X_train, y_train,
            epochs=10,
            batch_size=64,
            validation_split=0.2
        )
        
        histories[act] = history.history
        val_acc = history.history['val_accuracy'][-1]
        
        print(f"Activation: {act:5s} | Final Val Accuracy: {val_acc * 100:.2f}%")

plt.figure(figsize=(12, 5))

for act, hist in histories.items():
    plt.plot(hist['loss'], label=f'{act} (Train Loss)')
    plt.plot(hist['val_loss'], linestyle='--', label=f'{act} (Val Loss)')

plt.title('Training & Validation Loss Curves by Activation Function')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Answer
- ReLU (or GELU) performed best for this model, achieving slightly higher validation accuracy and faster convergence. A single experiment does not prove it is always better as it depend on network depth, dataset, and hyperparameter tuning.

## Task 10

```python
from keras.callbacks import EarlyStopping
from sklearn.metrics import confusion_matrix
import seaborn as sns

mlflow.set_experiment("Fashion_MNIST_Final_Evaluation")

with mlflow.start_run(run_name="Final_Best_Model"):
    mlflow.tensorflow.autolog()

    model = models.Sequential([
        layers.Flatten(input_shape=(28, 28)),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    model.fit(
        X_train, y_train,
        epochs=40,
        batch_size=32,
        validation_split=0.2,
        callbacks=[EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)]
    )

    test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
    print(f"\nFinal Test Accuracy: {test_acc * 100:.2f}%")


    predictions = model.predict(X_test)
    y_pred = np.argmax(predictions, axis=1)

    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names)
    plt.title('Confusion Matrix - Final Model')
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.tight_layout()
    plt.show()


    cm_no_diag = cm.copy()
    np.fill_diagonal(cm_no_diag, 0)
    i, j = np.unravel_index(np.argmax(cm_no_diag), cm_no_diag.shape)
    print(f"Most confused classes: '{class_names[i]}' often predicted as '{class_names[j]}' ({cm_no_diag[i, j]} times)")

    incorrect_indices = np.where(y_pred != y_test)[0][:3]

    plt.figure(figsize=(12, 4))
    for k, idx in enumerate(incorrect_indices):
        plt.subplot(1, 3, k + 1)
        plt.imshow(X_test[idx], cmap='gray')
        plt.title(f"True: {class_names[y_test[idx]]}\nPred: {class_names[y_pred[idx]]} ({predictions[idx][y_pred[idx]]*100:.1f}%)", color='red')
        plt.axis('off')
    plt.tight_layout()
    plt.show()
```

## Answer 
- Top clothing categories like Shirt, Pullover, Coat, and T-shirt/top are the hardest for the model because of their nearly identical visual silhouettes. To improve accuracy next, replace the fully-connected architecture with a CNN to extract spatial features like textures and edges.
