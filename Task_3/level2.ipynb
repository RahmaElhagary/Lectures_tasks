```python
def capitalize_list_items(lst):
    new_lst = []
    for item in lst:
        new_lst.append(item.capitalize())
    return new_lst

lst = ["rahma","mohammed","ahmed"]
capitalize_list_items(lst)
```

```python
def add_item(lst,item):
    lst.append(item)
    return lst

l = [1,2,4]
add_item(l,3)
print(l)
```

```python
def reverse(lst):
    r_lst = []
    for i in range(len(lst) - 1,-1,-1):
        r_lst.append(lst[i])
    return r_lst

l = [1,2,3,4,5]
print(reverse(l))
```

```python
def evens_and_odd(n):
    odd = even = 0
    for i in range(n + 1):
        odd += (i % 2 != 0)
        even += (i % 2 == 0)
    return (odd,even)

print(evens_and_odd(100))
```

```python
def calc_median(lst):
    sort_lst = sorted(lst)
    mid = sort_lst[len(lst) // 2]
    
    if len(lst) % 2 == 0:
        return (sort_lst[mid - 1] + sort_lst[mid]) / 2
    return sort_lst[mid]

def calc_mean(lst):
    return sum(lst) / len(lst)

from collections import Counter
def calc_mode(lst):
    c_lst = Counter(lst)
    return c_lst.most_common(1)[0][0]

lst = [1, 2, 3, 4, 4, 5]

print(calc_mode(lst))     
print(calc_mean(lst))  
print(calc_median(lst))
```

```python
def print_list(lst):
    for item in lst:
        print(item)
    
my_lst = [2,53,True]
print_list(my_lst)
```
