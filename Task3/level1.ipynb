```python
from math import pi
def area_of_circle(r):
    area = pi * r * r
    return area

def perimeter_of_area(r):
    parimeter = 2 * pi * r
    return parimeter
```

```python
def add_all_nums(*nums):
    if not all(isinstance(num,(int,float)) for num in nums):
        return
    return sum(nums)

print(add_all_nums(1,2,4))
```

```python
def convert_celsius_2_fahrenheit(c):
    f = (c * 9/5) + 32
    return f
```

```python
def check_season(month):
    month = month.strip().lower()
    if month in ['december', 'january', 'february']:
        return 'Winter'
    elif month in ['march', 'april', 'may']:
        return 'Spring'
    elif month in ['june', 'july', 'august']:
        return 'Summer'
    elif month in ['september', 'october', 'november']:
        return 'Autumn'
    
print(check_season("july"))
```

```python
def calculate_slope(x1,y1,x2,y2):
    m = y2 - y1 / x2 - x1
    return m
```

```python
def solve_quadratic_eqn(a,b,c):
    r1 = -b + (b**2 + 4*c*a)**0.5
    r2 = -b + (b**2 - 4*c*a)**0.5
    return r1,r2

solve_quadratic_eqn(1,-3,2)
```

```python
def print_list(lst):
    for item in lst:
        print(item)
    
my_lst = [2,53,True]
print_list(my_lst)
```

```python
def capitalize_list_items(lst):
    new_lst = []
    for item in lst:
        new_lst.append(item.capitalize())
    return new_lst

lst = ["rahma","mohammed","ahmed"]
capitalize_list_items(lst)
```

```python
def add_item(lst,item):
    lst.append(item)
    return lst

l = [1,2,4]
add_item(l,3)
print(l)
```

```python
def remove_item(lst,item):
    if item in lst:
        lst.remove(item)
    return lst

lst = [2,3,5,1]
remove_item(lst,1)
print(lst)
```

```python
def sum_of_numbers(n):
    sum = n * (n + 1) // 2
    return sum

print(sum_of_numbers(5))
print(sum_of_numbers(10))
print(sum_of_numbers(100))
```

```python
def sum_of_even(n): 
    sum = n // 2 * (n // 2 + 1)
    return sum

def sum_of_odd(n):
    sum = ((n + 1) // 2) ** 2
    return sum


print(sum_of_odd(7))
print(sum_of_even(8))
```
