```python
def is_prime(n):
    if n <= 1:
        return False
    
    for i in range(2,int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(2))
print(is_prime(6))
print(is_prime(1))
print(is_prime(5))
```

```python
def is_unique(lst):
    my_set = set(lst)
    return (len(my_set) == len(lst))

print(is_unique([1,1,12,4]))
print(is_unique([1,12,4]))
```

```python
def is_same_type(lst):
    types = set()
    for item in lst:
        types.add(type(item))
    
    return (len(types) == 1)

lst1 = [1,True,'color']
print(is_same_type(lst1))

lst2 = [1,2,3]
print(is_same_type(lst2))
```

```python
import keyword

def is_valid_variable(var_name):
    if not isinstance(var_name, str):
        return False
    return var_name.isidentifier() and not keyword.iskeyword(var_name)


print(is_valid_variable("my_var"))       
print(is_valid_variable("my var"))       
print(is_valid_variable("for"))
```
