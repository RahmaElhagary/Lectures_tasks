```python
country_land = []
for country in countriesWWWWW:
    if 'land' in country.lower():
        country_land.append(country)

print(country_land)
```

```python
lang = set()
for country in countries_data:
    for langs in country.get('languages',[]):
        lang.add(langs)
print(len(lang))
```

```python
from collections import Counter

all_lang = []
for country in countries_data:
    for lang in country.get('languages',[]):
        all_lang.append(lang)
        
lang_count = Counter(all_lang)
top_10_tuple = lang_count.most_common(10)  

top_10_lang = [lang[0] for lang in top_10_tuple]
print(top_10_lang)
```
